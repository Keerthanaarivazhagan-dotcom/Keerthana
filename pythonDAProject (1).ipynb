```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df["Engagement_Score"]=pd.to_numeric(df["Engagement_Score"], errors="coerce")

clean_df=df.dropna(subset=["Engagement_Score"])

plt.subplot(4, 1, 2)
sns.barplot(
    data=df,
    x="Channel_Used",
    y="Engagement_Score",
    hue="Campaign_Type",
    palette="muted",
    errorbar=None,
)
plt.xlabel("Channel Used",fontsize=12)
plt.ylabel("Engagaement Score",fontsize=12)
plt.legend(title="Campaign Type")
plt.title("Engagement Score by Channel Segment by Campaign Type")
```

```python
##Geographic Analysis:
## The highest Acquisition_Cost using a bar chart:
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df =pd.read_csv('https://raw.githubusercontent.com/ArchanaInsights/Datasets/main/marketing_campaign.csv')

cost_by_location = (
    df.groupby('Location')['Acquisition_Cost']
    .sum()
    .sort_values(ascending=False)
)
plt.figure(figsize=(10, 6))
cost_by_location.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Total Acquisition cost by location')
plt.xlabel('Location')
plt.ylabel('Acquisition Cost')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
pivot_conv = df.pivot_table(
    index='Location',
    columns='Target_Audience',
    values='Conversion_Rate',
    aggfunc='mean',
)
pivot_conv.plot(kind='bar', figsize=(12, 6), edgecolor='black')
plt.title('Average Conversion Rate by Location and Target Audience')
plt.xlabel('Location')
plt.ylabel('Conversion Rate')
plt.legend(title='Target Audience')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## Using a Pie chart:
```

```python
roi_by_Location = df.groupby('Location')['ROI'].sum()

plt.figure(figsize=(8, 8))
plt.pie(
    roi_by_Location,
    labels=roi_by_Location.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Set3.colors,
)

plt.title('Proportion of Total ROI by Location')
plt.axis('equal')
plt.tight_layout()
plt.show()
```
